**2 - LOAD DATA**

```python
path_label = "/content/drive/MyDrive/SKRIPSI/CODE/HASIL_PREPROCESSING"

df_a1 = pd.read_csv(f"{path_label}/anotator1.csv")
df_a2 = pd.read_csv(f"{path_label}/anotator2.csv")

print(f"✅ Anotator 1 : {len(df_a1)} data")
print(f"✅ Anotator 2 : {len(df_a2)} data")
```

**3 - AMBIL 100 DATA PERTAMA**

```python
df_a1_100 = df_a1.head(100).copy()
df_a2_100 = df_a2.head(100).copy()
```

**4 - VALIDASI DATA**

```python
assert len(df_a1_100) == len(df_a2_100), "❌ Jumlah data tidak sama!"
```

**5 - BERSIHKAN LABEL**

```python
df_a1_100['label'] = df_a1_100['label'].astype(str).replace(['', ' '], '0')
df_a2_100['label'] = df_a2_100['label'].astype(str).replace(['', ' '], '0')
```

**6 - HITUNG COHEN'S KAPPA**

```python
kappa = cohen_kappa_score(df_a1_100['label'], df_a2_100['label'])

print(f"\n✅ Cohen's Kappa (κ): {kappa:.4f}\n")

if kappa < 0.21:
    print("None")
elif kappa < 0.40:
    print("Minimal Agreement")
elif kappa < 0.60:
    print("Weak Agreement")
elif kappa < 0.80:
    print("Moderate Agreement")
elif kappa <= 0.90:
    print("Strong Agreement")
else:
    print("Almost Perfect Agreement")
```

**7- HITUNG Po & Pe**

```python
from sklearn.metrics import confusion_matrix
import numpy as np

cm = confusion_matrix(df_a1_100['label'], df_a2_100['label'])

# Observed agreement (Po)
Po = np.trace(cm) / np.sum(cm)

# Expected agreement (Pe)
row_sum = np.sum(cm, axis=1)
col_sum = np.sum(cm, axis=0)
Pe = np.sum(row_sum * col_sum) / (np.sum(cm) ** 2)

print("Po:", Po)
print("Pe:", Pe)

# Hitung manual kappa
kappa_manual = (Po - Pe) / (1 - Pe)
print("Kappa (manual):", kappa_manual)
```
